## Parte 1 - Pensamento computacional

**1. Chamados aguardando atendimento**

Vou usar uma fila, porque os chamados precisam ser atendidos na ordem em que chegaram. Isso é FIFO (o primeiro que entra é o primeiro que sai).

**2. Chamados que já foram atendidos**

Vou usar uma pilha, porque quando precisar desfazer, tenho que pegar o último atendimento primeiro. Isso é LIFO.

**3. Simulação**

No começo, a fila tem os chamados 101, 102, 103 e 104.

- Primeiro `atender`: sai o 101. Fila: 102, 103, 104. Atendidos: 101.
- Segundo `atender`: sai o 102. Fila: 103, 104. Atendidos: 101, 102.
- `desfazer`: o 102 volta para o começo da fila. Fila: 102, 103, 104. Atendidos: 101.
- Último `atender`: sai o 102 de novo. Fila: 103, 104. Atendidos: 101, 102.

A prioridade fica guardada no chamado, mas nessa atividade o atendimento segue a ordem de chegada.

## Parte 2 - Implementação

A classe `CentralAtendimento` tem uma fila para quem está esperando e uma pilha para guardar os atendimentos feitos.

```python
from collections import deque

class CentralAtendimento:
    def __init__(self):
        self.fila_espera = deque()
        self.chamados_atendidos = []  # pilha

    def adicionar_chamado(self, codigo, cliente, prioridade):
        chamado = {
            'codigo': codigo,
            'cliente': cliente,
            'prioridade': prioridade
        }
        self.fila_espera.append(chamado)
        return chamado

    def atender_proximo(self):
        if not self.fila_espera:
            print('Não há chamados aguardando atendimento.')
            return None
        chamado = self.fila_espera.popleft()
        self.chamados_atendidos.append(chamado)
        print(f\"Atendido: {chamado['codigo']} - {chamado['cliente']}\")
        return chamado

    def desfazer_ultimo_atendimento(self):
        if not self.chamados_atendidos:
            print('Não há atendimento para desfazer.')
            return None
        chamado = self.chamados_atendidos.pop()
        self.fila_espera.appendleft(chamado)
        print(f\"Atendimento desfeito: {chamado['codigo']} - {chamado['cliente']}\")
        return chamado

    def buscar_chamado(self, codigo):
        for chamado in self.fila_espera:
            if chamado['codigo'] == codigo:
                return chamado
        return None

    def exibir_situacao(self):
        aguardando = [c['codigo'] for c in self.fila_espera]
        atendidos = [c['codigo'] for c in self.chamados_atendidos]
        print(f'Chamados aguardando: {aguardando}')
        print(f'Chamados atendidos: {atendidos}')
```

## Parte 3 - Teste

Agora vou executar a sequência que está no enunciado.

```python
central = CentralAtendimento()

central.adicionar_chamado(101, 'Ana', 'Média')
central.adicionar_chamado(102, 'Bruno', 'Alta')
central.adicionar_chamado(103, 'Carla', 'Baixa')
central.adicionar_chamado(104, 'Daniel', 'Alta')

central.atender_proximo()
central.atender_proximo()
central.desfazer_ultimo_atendimento()
central.atender_proximo()
central.exibir_situacao()
print(central.buscar_chamado(104))
```

## Resultado

Depois dos comandos, devem ficar aguardando os chamados **103 e 104**. Os chamados **101 e 102** ficam na lista de atendidos.

A busca pelo código 104 retorna os dados do Daniel, porque ele ainda está na fila.

## Parte 4 - Análise

A busca tem complexidade **O(n)** no pior caso, porque pode precisar passar por todos os chamados da fila até encontrar o código.
