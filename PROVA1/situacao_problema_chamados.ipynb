## Sistema de controle de chamados de suporte

### Parte 1 — Pensamento computacional

**1. Estrutura para chamados aguardando:** fila (`deque`), pois os chamados devem ser atendidos na ordem de chegada (**FIFO — First In, First Out**).

**2. Estrutura para chamados atendidos:** pilha (lista Python), pois o sistema precisa desfazer primeiro o atendimento mais recente (**LIFO — Last In, First Out**).

**3. Simulação antes de executar o código:**

| Operação | Fila de espera | Atendidos (base → topo) |
|---|---|---|
| Inicial | 101, 102, 103, 104 | — |
| atender | 102, 103, 104 | 101 |
| atender | 103, 104 | 101, 102 |
| desfazer | 102, 103, 104 | 101 |
| atender | 103, 104 | 101, 102 |

A prioridade é armazenada em cada chamado, mas a regra fornecida determina atendimento pela ordem de chegada; portanto, não reorganizamos a fila por prioridade.

### Parte 2 — Implementação
A classe mantém uma fila de espera e uma pilha de atendimentos. A busca percorre apenas os chamados que ainda aguardam atendimento.

```python
from collections import deque

class CentralAtendimento:
    def __init__(self):
        self.fila_espera = deque()
        self.chamados_atendidos = []  # pilha: o último é o topo

    def adicionar_chamado(self, codigo, cliente, prioridade):
        chamado = {
            'codigo': codigo,
            'cliente': cliente,
            'prioridade': prioridade
        }
        self.fila_espera.append(chamado)
        return chamado

    def atender_proximo(self):
        if not self.fila_espera:
            print('Não há chamados aguardando atendimento.')
            return None
        chamado = self.fila_espera.popleft()
        self.chamados_atendidos.append(chamado)
        print(f"Atendido: {chamado['codigo']} - {chamado['cliente']}")
        return chamado

    def desfazer_ultimo_atendimento(self):
        if not self.chamados_atendidos:
            print('Não há atendimento para desfazer.')
            return None
        chamado = self.chamados_atendidos.pop()
        self.fila_espera.appendleft(chamado)
        print(f"Atendimento desfeito: {chamado['codigo']} - {chamado['cliente']}")
        return chamado

    def buscar_chamado(self, codigo):
        for chamado in self.fila_espera:
            if chamado['codigo'] == codigo:
                return chamado
        return None

    def exibir_situacao(self):
        aguardando = [c['codigo'] for c in self.fila_espera]
        atendidos = [c['codigo'] for c in self.chamados_atendidos]
        print(f'Chamados aguardando: {aguardando}')
        print(f'Chamados atendidos: {atendidos}')
```

### Parte 3 — Teste obrigatório
Execute a sequência abaixo conforme o enunciado.

```python
central = CentralAtendimento()

central.adicionar_chamado(101, 'Ana', 'Média')
central.adicionar_chamado(102, 'Bruno', 'Alta')
central.adicionar_chamado(103, 'Carla', 'Baixa')
central.adicionar_chamado(104, 'Daniel', 'Alta')

central.atender_proximo()
central.atender_proximo()
central.desfazer_ultimo_atendimento()
central.atender_proximo()
central.exibir_situacao()
print(central.buscar_chamado(104))
```

### Resultado esperado
- Atendido inicialmente: 101 (Ana).
- Segundo atendimento: 102 (Bruno).
- Desfazer devolve 102 para o início da fila.
- O próximo atendimento é novamente 102 (Bruno).
- Fila final: `[103, 104]`.
- Atendidos finais: `[101, 102]`.
- `buscar_chamado(104)` retorna os dados de Daniel, pois 104 ainda está aguardando.

### Parte 4 — Análise de complexidade
No pior caso, `buscar_chamado()` tem complexidade **O(n)**, pois precisa percorrer sequencialmente até os n chamados da fila para encontrar o código ou concluir que ele não existe.
